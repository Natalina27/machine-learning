## Supervised learning (обучение с учителем)

Это самый частый тип ML. У нас есть **размеченные данные** — для каждого примера мы знаем и вход, и правильный ответ. Модель смотрит на пары «вход → ответ» и учится находить закономерность: *если признаки такие, то результат, скорее всего, такой*.

Программист не пишет правила вручную через `if/else` — модель сама подстраивает параметры по примерам. После обучения её можно применить к **новым** данным, где ответа ещё нет.

Противоположность — **unsupervised learning** (без учителя): там меток нет, модель ищет структуру в данных сама (кластеры, аномалии и т.д.).

## Классификация vs регрессия

Обе задачи — supervised, но **тип ответа** разный:

| | Классификация | Регрессия |
|---|---------------|-----------|
| **Что предсказываем** | категорию / класс | число |
| **Примеры** | спам / не спам, выжил / нет, тип ириса | цена квартиры, зарплата, температура |
| **y** | дискретное (0/1, A/B/C) | непрерывное (любое число) |

На практике: если ответ — «к какой группе относится объект?», это классификация. Если «сколько?» — регрессия.

## X и y

Перед обучением таблицу нужно разделить на две части:

- **X (features, признаки)** — то, что модель **видит на входе**. Столбцы с описанием объекта: возраст, доход, площадь квартиры.
- **y (target, цель)** — то, что модель **должна предсказать**. Один столбец с ответом: купил / не купил, цена, диагноз.

Каждая **строка** — один объект (одно наблюдение). Модель учится: *по строке X угадать значение y*.

В коде это обычно выглядит так: `X = df[["age", "income"]]`, `y = df["bought"]`.

## Train / test split

Данные делят на две части:

- **Train (обучающая выборка)** — на ней модель **учится** (`fit`).
- **Test (тестовая выборка)** — модель **не видела** эти строки при обучении; на них проверяют качество (`predict` + метрики).

Зачем: если проверять модель на тех же данных, на которых она училась, метрики будут **завышены** — модель могла просто «зазубрить» ответы (overfitting), а на новых данных провалится.

Типичное разбиение: 80% train / 20% test. Важно: split делают **до** любой предобработки (масштабирование, заполнение пропусков), иначе возможен leakage.

## Data leakage (утечка данных)

**Утечка** — когда в обучение попадает информация, которой модель **не должна быть доступна** в реальной задаче. Метрики на test выглядят отлично, а на новых данных модель бесполезна.

Типичные случаи:

- **Масштабирование / imputation на всём датасете** до split — test «подсмотрел» статистику train через общий scaler.
- **Признак, который напрямую содержит ответ** — например, «статус сделки = купил» при предсказании покупки.
- **Информация из будущего** — предсказываем продажи за январь, а в признаках есть данные за февраль.

**Правило:** всё, что «узнаёт» что-то из данных (среднее, медиана, частоты), должно считаться **только на train** и применяться к test уже обученным преобразователем. Для этого используют `Pipeline` — об этом подробнее в день 6.

## 4. Три вопроса по датасету

1. **Что предсказываем?** Вид ириса (target): setosa, versicolor, virginica — задача классификации.

## Supervised learning

This is the most common type of ML. We have **labeled data** — for each example we know both the input and the correct answer. The model looks at "input → answer" pairs and learns the pattern: *if the features look like this, the outcome is probably that*.

The programmer does not write rules manually with `if/else` — the model adjusts its parameters from examples. After training, it can be applied to **new** data where the answer is not yet known.

The opposite is **unsupervised learning**: there are no labels, and the model finds structure in the data on its own (clusters, anomalies, etc.).

## Classification vs regression

Both tasks are supervised, but the **type of answer** differs:

| | Classification | Regression |
|---|----------------|------------|
| **What we predict** | category / class | a number |
| **Examples** | spam / not spam, survived / not, iris species | apartment price, salary, temperature |
| **y** | discrete (0/1, A/B/C) | continuous (any number) |

In practice: if the question is "which group does this object belong to?", it's classification. If it's "how much?", it's regression.

## X and y

Before training, split the table into two parts:

- **X (features)** — what the model **sees as input**. Columns describing the object: age, income, apartment size.
- **y (target)** — what the model **must predict**. One column with the answer: bought / not bought, price, diagnosis.

Each **row** is one object (one observation). The model learns: *given a row of X, guess the value of y*.

In code this usually looks like: `X = df[["age", "income"]]`, `y = df["bought"]`.

## Train / test split

Split the data into two parts:

- **Train (training set)** — the model **learns** on this data (`fit`).
- **Test (test set)** — the model **has not seen** these rows during training; quality is checked on them (`predict` + metrics).

Why: if you evaluate on the same data the model trained on, metrics will be **inflated** — the model may have simply memorized the answers (overfitting) and will fail on new data.

Typical split: 80% train / 20% test. Important: split **before** any preprocessing (scaling, imputation), otherwise leakage is possible.

## Data leakage

**Leakage** happens when information that the model **should not have access to** in a real task gets into training. Test metrics look great, but the model is useless on new data.

Typical cases:

- **Scaling / imputation on the full dataset** before the split — test "peeked" at train statistics through a shared scaler.
- **A feature that directly contains the answer** — e.g. "deal status = bought" when predicting a purchase.
- **Information from the future** — predicting January sales while features include February data.

**Rule:** anything that "learns" from data (mean, median, frequencies) must be computed **on train only** and applied to test with the fitted transformer. Use a `Pipeline` for this — more on that in Day 6.
