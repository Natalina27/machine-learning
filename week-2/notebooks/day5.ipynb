## Работа с файлами
### open/read/write, кодировка UTF‑8

**Режимы открытия:**
- `open("file.txt", "r")` — открыть для чтения
- `open("file.txt", "w")` — открыть для записи (перезаписывает!)
- `open("file.txt", "a")` — добавить в конец
- `with open(...) as f:` — автоматически закрывает файл
- `encoding="utf-8"` — поддержка всех языков включая русский

**Методы чтения:**
- `f.read()` — весь файл как одна строка
- `f.readlines()` — список строк
- `f.readline()` — одна строка

**Методы записи:**
- `f.write("текст")` — записать строку
- `f.writelines(["строка1\n", "строка2\n"])` — записать список строк

### pathlib — работа с путями

- `Path(".")` — текущая папка
- `Path("data") / "file.csv"` — объединить пути (вместо строк)
- `path.exists()` — проверить существование
- `path.mkdir(exist_ok=True)` — создать папку
- `path.iterdir()` — список файлов в папке
- Преимущество: работает на всех ОС (не нужно думать о `/` или `\`)

```python
from pathlib import Path

# Ищем data-папку независимо от cwd.
# В этом проекте sample.txt и utils.py лежат в week-2/notebooks/data.
DATA_CANDIDATES = [
    Path("week-2/notebooks/data"),  # запуск из корня репо
    Path("data"),                   # запуск из week-2/notebooks/
    Path.cwd(),                      # fallback: если файлы рядом с текущей папкой
]

DATA_DIR = next((p for p in DATA_CANDIDATES if (p / "sample.txt").exists()), None)
if DATA_DIR is None:
    raise FileNotFoundError(
        "Не найден sample.txt. Ожидался в week-2/notebooks/data или data."
    )

SAMPLE_FILE = DATA_DIR / "sample.txt"
RESULTS_FILE = DATA_DIR / "results.txt"

print(f"Текущая папка (cwd): {Path.cwd().resolve()}")
print(f"Папка data: {DATA_DIR.resolve()}")
print(f"sample.txt: {SAMPLE_FILE} — exists: {SAMPLE_FILE.exists()}")

print("\nФайлы в data:")
for f in sorted(DATA_DIR.iterdir()):
    if f.is_file():
        print(f"  {f.name}")
```

## Задания

```python
# Взять текстовый файл (лог/статья) и посчитать: 
# количество строк, уникальные слова, топ‑10 слов
from collections import Counter

with open(SAMPLE_FILE, "r", encoding="utf-8") as f:
    words = f.read().split()

# Топ-10 слов
top_words = Counter(words).most_common(10)
print("Топ-10 слов:")
for word, count in top_words:
    print(f"  {word}: {count}")

unique_words = set(words)
print(f"Уникальных слов: {len(unique_words)}")

# количество строк
with open(SAMPLE_FILE, "r", encoding="utf-8") as f:
    lines = f.readlines()

print(f"Количество строк: {len(lines)}")
```

```python
# Сохраняем результат в файл
with open(RESULTS_FILE, "w", encoding="utf-8") as f:
    f.write(f"Количество строк: {len(lines)}\n")
    f.write(f"Уникальных слов: {len(unique_words)}\n\n")
    f.write("Топ-10 слов:\n")
    for word, count in top_words:
        f.write(f"{word}: {count}\n")

print(f"Результаты сохранены в {RESULTS_FILE}")
```

## Working with files
### open/read/write, UTF-8 encoding

**Open modes:**
- `open("file.txt", "r")` — read
- `open("file.txt", "w")` — write (overwrites!)
- `open("file.txt", "a")` — append
- `with open(...) as f:` — closes the file automatically
- `encoding="utf-8"` — supports all languages including non-ASCII

**Read methods:**
- `f.read()` — entire file as one string
- `f.readlines()` — list of lines
- `f.readline()` — one line

**Write methods:**
- `f.write("text")` — write a string
- `f.writelines(["line1\n", "line2\n"])` — write a list of strings


### pathlib — working with paths

- `Path(".")` — current directory
- `Path("data") / "file.csv"` — join paths (instead of strings)
- `path.exists()` — check existence
- `path.mkdir(exist_ok=True)` — create directory
- `path.iterdir()` — list files in a directory
- Advantage: works on all OS (no need to worry about `/` or `\`)
