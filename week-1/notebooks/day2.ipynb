## Очистка данных

ML не работает с пропусками «как есть». Стратегии:

- **Age** — много пропусков → заполняем **медианой** (устойчива к выбросам)
- **Embarked** — мало пропусков → заполняем **самым частым** значением (`mode`)
- **Cabin** — ~77% пропусков → колонку **удаляем**, информации мало

После чистки снова проверяем `isnull().sum()` — должно быть 0 (кроме колонок, которые мы оставили намеренно).

````python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop(columns=['Cabin'])

df.isnull().sum()
````

## Кодирование категорий

Модели принимают **числа**, не текст. Колонка `Sex` — `male` / `female` → кодируем в 0 / 1 через `.map()`:

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

Это бинарная кодировка. Другие категории (например `Embarked`) кодируют позже — one-hot или label encoding.

````python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
df['Sex'].head()
````

## describe() — числовая статистика

`describe()` по числовым колонкам: count, mean, std, min, 25%/50%/75%, max.

Помогает заметить выбросы (например, Fare до 512) и понять масштаб признаков.

````python
df.describe()
````

## Выводы

- `head()`, `info()`, `isnull().sum()` — базовый осмотр перед ML.
- Заполнили пропуски в Age и Embarked, удалили Cabin.
- Закодировали `Sex` в числа — модель сможет использовать этот признак.
- `describe()` показал распределение числовых колонок.

# Day 2 — Pandas basics

## Goal
Learn to load CSV, inspect data, and clean it — the first step before any ML model.

Dataset: **Titanic** — 891 passengers, 12 columns. Later we predict survival; today is exploration and prep only.

## Loading data

- `pd.read_csv(path)` — reads a table from a CSV file
- **`df`** (DataFrame) — main pandas structure: rows = objects, columns = features
- Path `../../data/titanic.csv` — from `week-1/notebooks/` to `data/` at repo root

## Initial inspection

Always look inside before modeling:

| Method | Purpose |
|--------|--------|
| `head()` | first 5 rows — what the data looks like |
| `info()` | column types, non-null counts, memory |
| `isnull().sum()` | missing values per column |

## Data cleaning

ML cannot use missing values as-is. Strategies:

- **Age** — many gaps → fill with **median** (robust to outliers)
- **Embarked** — few gaps → fill with **most frequent** value (`mode`)
- **Cabin** — ~77% missing → **drop** the column, too little information

After cleaning, check `isnull().sum()` again — should be 0 for columns you kept.

## Encoding categories

Models need **numbers**, not text. Column `Sex` — `male` / `female` → encode as 0 / 1 with `.map()`:

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

This is binary encoding. Other categories (e.g. `Embarked`) are encoded later — one-hot or label encoding.

## describe() — numeric statistics

`describe()` on numeric columns: count, mean, std, min, 25%/50%/75%, max.

Helps spot outliers (e.g. Fare up to 512) and understand feature scale.

## Conclusions

- Loaded Titanic with `pd.read_csv()` — DataFrame 891 × 12.
- `head()`, `info()`, `isnull().sum()` — basic inspection before ML.
- Filled missing Age and Embarked, dropped Cabin.
- Encoded `Sex` as numbers — the model can use this feature.
- `describe()` showed distribution of numeric columns.

Data is ready for the next step: pick X and y and train a model (day 4+).
